```python
import pandas as pd
```

```python
df_colunas = pd.read_csv('colunas_variaveis.csv', sep=';')
```

```python
# Lista de colunas para analisar
colunas_analisar = [
    'TIPOBITO', 'DTOBITO', 'DTNASC', 'IDADE', 'SEXO', 'RACACOR', 'ESTCIV',
    'ESC2010', 'CODMUNRES', 'LOCOCOR', 'CODESTAB', 'CODMUNOCOR', 'IDADEMAE',
    'QTDFILVIVO', 'QTDFILMORT', 'SEMAGESTAC', 'GRAVIDEZ', 'PARTO',
    'OBITOPARTO', 'TPMORTEOCO', 'ASSISTMED', 'LINHAA', 'LINHAB', 'LINHAC',
    'LINHAD', 'LINHAII', 'CAUSABAS', 'OBITOGRAV', 'OBITOPUERP', 'CAUSABAS_O',
    'DTINVESTIG', 'DTCADASTRO', 'FONTEINV', 'DTRECEBIM', 'CAUSAMAT',
    'NUDIASOBCO', 'DTCADINV', 'TPOBITOCOR', 'DTCONINV', 'TPRESGINFO',
    'TPNIVELINV', 'DTCADINF', 'MORTEPARTO', 'ALTCAUSA', 'TPPOS', 'TP_ALTERA'
]

def verificar_colunas_sim(df, colunas):
    """
    Verifica em quais anos cada coluna está disponível (True) nos registros
    onde Fonte = 'SIM'.
    
    Parâmetros:
        df: DataFrame com as colunas Ano, Fonte e as variáveis booleanas
        colunas: lista de nomes de colunas a verificar
    
    Retorna:
        DataFrame com índice = colunas, colunas = anos, valores = True/False
    """
    df_sim = df[df['Fonte'] == 'SIM'].copy()
    anos = sorted(df_sim['Ano'].unique())
    
    resultados = {}
    for col in colunas:
        if col in df_sim.columns:
            resultados[col] = {ano: df_sim.loc[df_sim['Ano'] == ano, col].values[0]
                               for ano in anos}
        else:
            resultados[col] = {ano: False for ano in anos}
    
    df_resultado = pd.DataFrame(resultados).T
    df_resultado.index.name = 'Coluna'
    return df_resultado


# Executar a verificação
df_resultado = verificar_colunas_sim(df_colunas, colunas_analisar)

# Exibir resultado
print("Colunas disponíveis por ano (Fonte = SIM):")
display(df_resultado)

# Mostrar apenas os anos em que cada coluna está presente
print("\n--- Resumo: anos em que cada coluna está disponível ---")
for coluna in colunas_analisar:
    anos_presentes = df_resultado.columns[df_resultado.loc[coluna] == True].tolist()
    if anos_presentes:
        print(f"{coluna}: {anos_presentes}")
    else:
        print(f"{coluna}: NÃO ENCONTRADA em nenhum ano do SIM")
```
